```python
#Obtain Epoch Timestamp on bl132lx
colnames=['Key','C0','C1','C2','C3','C4','C5',
          'C6','C7','C8','C9','C10',
          'C11','C12','C13','C14','C15',
          'C16','C17','C18','C19','C20']
colnames_sample=['Sample_C','C0','C1','SAMPLE','leftbot_hor','leftbot_dep','leftbot_ver','righttop_hor','righttop_dep','righttop_ver','C2','C3']

for _experiment in experiment:
    inputfile=home+_experiment+'/htxssample.csv'
    df_sample = pd.read_csv(inputfile, names=colnames_sample, header=None,sep=' ')
    df_sample['ID']=df_sample['C0']*100+df_sample['C1']
    df_sample['R_min']=((df_sample['C0'])%4*90-90+0.1)%360
    df_sample['R_max']=((df_sample['C0'])%4*90+0.1)%360

    df_sample.loc[(df_sample['C0']==1),'R_cenX'] = R_cen[0][0]
    df_sample.loc[(df_sample['C0']==1),'R_cenY'] = R_cen[0][1]
    df_sample.loc[(df_sample['C0']==2),'R_cenX'] = R_cen[1][0]
    df_sample.loc[(df_sample['C0']==2),'R_cenY'] = R_cen[1][1]
    df_sample.loc[(df_sample['C0']==3),'R_cenX'] = R_cen[2][0]
    df_sample.loc[(df_sample['C0']==3),'R_cenY'] = R_cen[2][1]
    df_sample.loc[(df_sample['C0']==4),'R_cenX'] = R_cen[3][0]
    df_sample.loc[(df_sample['C0']==4),'R_cenY'] = R_cen[3][1]
    df_sample.loc[(df_sample['C0']==1),'Z_min'] = R_cen[0][2]+df_sample['leftbot_ver']
    df_sample.loc[(df_sample['C0']==1),'Z_max'] = R_cen[0][2]+df_sample['righttop_ver']
    df_sample.loc[(df_sample['C0']==2),'Z_min'] = R_cen[1][2]+df_sample['leftbot_ver']
    df_sample.loc[(df_sample['C0']==2),'Z_max'] = R_cen[1][2]+df_sample['righttop_ver']
    df_sample.loc[(df_sample['C0']==3),'Z_min'] = R_cen[2][2]+df_sample['leftbot_ver']
    df_sample.loc[(df_sample['C0']==3),'Z_max'] = R_cen[2][2]+df_sample['righttop_ver']
    df_sample.loc[(df_sample['C0']==4),'Z_min'] = R_cen[3][2]+df_sample['leftbot_ver']
    df_sample.loc[(df_sample['C0']==4),'Z_max'] = R_cen[3][2]+df_sample['righttop_ver']


    df_sample['X_cen']=((df_sample['leftbot_hor']+df_sample['righttop_hor'])-stick_width)/2
    df_sample['X_min']=df_sample['leftbot_hor']-stick_width/2
    df_sample['X_max']=df_sample['righttop_hor']-stick_width/2
    df_sample['Y_cen']=(df_sample['leftbot_dep']+df_sample['righttop_dep'])/2

    df_sample.loc[(df_sample['C0']==1),'Z_cen'] = R_cen[0][2]+(df_sample['leftbot_ver']+df_sample['righttop_ver'])/2
    df_sample.loc[(df_sample['C0']==2),'Z_cen'] = R_cen[1][2]+(df_sample['leftbot_ver']+df_sample['righttop_ver'])/2
    df_sample.loc[(df_sample['C0']==3),'Z_cen'] = R_cen[2][2]+(df_sample['leftbot_ver']+df_sample['righttop_ver'])/2
    df_sample.loc[(df_sample['C0']==4),'Z_cen'] = R_cen[3][2]+(df_sample['leftbot_ver']+df_sample['righttop_ver'])/2

    #RR=15
    #df_sample['X_cen']=(df_sample['X_cen'])*math.sin(RR/180*math.pi)-np.sqrt(2)*(stick_width/2-df_sample['Y_cen'])*math.cos((90+45-RR)/180*math.pi)  +df_sample['R_cenX']*math.cos((90+45-RR)/180*math.pi) #df_sample['R_cenX']#+df_sample['R_cenX']#-math.sin(15/180*math.pi)*stick_width+SampleX_min
    #df_sample['X_min']=(df_sample['X_min'])*math.sin(RR/180*math.pi)-np.sqrt(2)*(stick_width/2-df_sample['Y_cen'])*math.cos((90+45-RR)/180*math.pi)  +df_sample['R_cenX']*math.cos((90+45-RR)/180*math.pi) #df_sample['R_cenX']#+df_sample['R_cenX']#-math.sin(15/180*math.pi)*stick_width+SampleX_min
    #df_sample['X_max']=(df_sample['X_max'])*math.sin(RR/180*math.pi)-np.sqrt(2)*(stick_width/2-df_sample['Y_cen'])*math.cos((90+45-RR)/180*math.pi)  +df_sample['R_cenX']*math.cos((90+45-RR)/180*math.pi) #df_sample['R_cenX']#+df_sample['R_cenX']#-math.sin(15/180*math.pi)*stick_width+SampleX_min


    df_sample['E']=_experiment
    outfile=dir+'/pandas/df_sample.pkl'
    df_sample.to_pickle(outfile)


    inputfile=dir+'/htxscom.csv'
    df_comments = pd.read_csv(inputfile, names=colnames, header=None,sep= ' ',na_filter=False)

    df_epoch0=df_comments[(df_comments['Key']=='#E')]['C0']
    df_epoch0_list=df_epoch0.to_list()
    df_epoch0index_list=df_epoch0.index.to_list()
    Epoch0=float(df_comments[(df_comments['Key']=='#E')]['C0'].values[0])-Epoch_offset

    df_sn=df_comments[(df_comments['Key']=='#S')]['C0'].reset_index().drop(columns=['index']).rename(columns={"C0": "S"})
    df_x=df_comments[(df_comments['Key']=='#P1')]['C6'].reset_index().drop(columns=['index']).rename(columns={"C6": "X"})
    df_y=df_comments[(df_comments['Key']=='#P1')]['C7'].reset_index().drop(columns=['index']).rename(columns={"C7": "Y"})
    df_z=df_comments[(df_comments['Key']=='#P2')]['C0'].reset_index().drop(columns=['index']).rename(columns={"C0": "Z"})
    df_r=df_comments[(df_comments['Key']=='#P2')]['C1'].reset_index().drop(columns=['index']).rename(columns={"C1": "R"})
    df_gap=df_comments[(df_comments['Key']=='#P0')]['C1'].reset_index().drop(columns=['index']).rename(columns={"C1": "GAP"})
    df_epu=df_comments[(df_comments['Key']=='#P0')]['C0'].reset_index().drop(columns=['index']).rename(columns={"C0": "EPU"})
    df_sgmensw=df_comments[(df_comments['Key']=='#P0')]['C5'].reset_index().drop(columns=['index']).rename(columns={"C5": "SGMENSW"})
    df_sgmensz=df_comments[(df_comments['Key']=='#P0')]['C4'].reset_index().drop(columns=['index']).rename(columns={"C4": "SGMENSZ"})
    df_m1pit=df_comments[(df_comments['Key']=='#P0')]['C3'].reset_index().drop(columns=['index']).rename(columns={"C3": "M1PIT"})
    df_m0pit=df_comments[(df_comments['Key']=='#P0')]['C2'].reset_index().drop(columns=['index']).rename(columns={"C2": "M0PIT"})
    df_exsv_2=df_comments[(df_comments['Key']=='#P1')]['C0'].reset_index().drop(columns=['index']).rename(columns={"C0": "EXSV_2"})
    df_sgmhorz=df_comments[(df_comments['Key']=='#P0')]['C6'].reset_index().drop(columns=['index']).rename(columns={"C6": "SGMHORZ"})
    df_mono132=df_comments[(df_comments['Key']=='#P1')]['C1'].reset_index().drop(columns=['index']).rename(columns={"C1": "MONO132"})

    droplist=df_comments.loc[(df_comments['Key']=='#E')].index.to_list()
    i=0
    for y in droplist:
        droplist[i]=y+1
        i=i+1
    df_comments['EPOCH0']=df_epoch0_list[0]
    kk=0
    for k in df_epoch0index_list:
        df_comments.loc[(df_comments.index>=k),'EPOCH0'] = df_epoch0_list[kk]
        kk=kk+1

    df_datetime=df_comments.loc[(df_comments['Key']=='#D')][["C4","C1","C2","C3","EPOCH0"]].drop(index=droplist).reset_index().drop(columns=['index'])

    df_datetime['DATETIME'] = df_datetime.agg(lambda x: f"{x['C4']} {x['C1']} {x['C2']} {x['C3']}", axis=1)

    df_datetime['EPOCH']=''
    for index, row in df_datetime.iterrows():
        row['EPOCH'] = datetime.datetime.timestamp(datetime.datetime.strptime(row['DATETIME'],"%Y %b %d %H:%M:%S"))-Epoch_offset
    df_datetime['EPOCH']=row['EPOCH']
    df_epoch=df_datetime['EPOCH'].reset_index().drop(columns=['index'])
    df_epoch0=df_datetime['EPOCH0'].reset_index()#.drop(columns=['index'])


    df_scan =df_comments[df_comments['Key']=='#S'].reset_index().drop(columns=['index'])   # key to extract fascan

    df_scan['colnames']=''
    df_scan['scan']=''
    colnames0="'"+df_scan["C2"]+"'"
    colnames1="'mono132','gap'"
    colnames2="'time'"
    df_scan.loc[((df_scan['C1']=='gscan') | (df_scan['C1']=='ascan')),'colnames']=colnames0
    df_scan.loc[((df_scan['C1']=='tes_local_gscan') | (df_scan['C1']=='ascan')),'colnames']=colnames0
    df_scan.loc[(df_scan['C1']=='fscan'),'colnames']=colnames1
    df_scan.loc[(df_scan['C1']=='timescan')| (df_scan['C1']=='loopscan'),'colnames']=colnames2

    scancmd_list=[]
    for index, row in df_scan.iterrows():
        scancmd=f"{row['C1']} {row['C2']} {row['C3']} {row['C4']} {row['C5']} {row['C6']} {row['C7']} {row['C8']} {row['C9']} {row['C10']} {row['C11']} {row['C12']}"

        scancmd_list.append(scancmd.strip())


    df_scancmd=pd.DataFrame(scancmd_list, columns=['scancmd'])
    colname_list=df_scan['colnames'].tolist()
    df_colnames=pd.DataFrame(colname_list, columns=['C'])

    df_xyzr= pd.concat([df_sn,df_x,df_y,df_z,df_r,df_mono132,df_gap,df_epu,df_m0pit,df_m1pit,df_sgmensw,df_sgmensz,df_sgmhorz,df_exsv_2,df_epoch,df_scancmd,df_colnames,df_epoch0], axis=1)

    samplename_list=[]
    sampleid_list=[]
    scanstring_list=[]
    angle_list=[]
    for index, row in df_xyzr.iterrows():
        X0=round(float(row['X']),2)
        Y0=round(float(row['Y']),2)
        Z0=round(float(row['Z']),2)
        R0=round(float(row['R']),2)
        M0=round(float(row['MONO132']),2)

        samplename=samplename_from_xyzr(df_sample,X0,Y0,Z0,R0)
        sampleid=sampleid_from_xyzr(df_sample,X0,Y0,Z0,R0)
        samplename_list.append(samplename)
        angle_list.append(round(R0%90,1))

        sampleid_list.append(sampleid)

        scanstring_list.append(scanstring_from_mono132(M0))

    df_angle=pd.DataFrame(angle_list, columns=['angle']).reset_index().drop(columns=['index'])
    df_samplename=pd.DataFrame(samplename_list, columns=['samplename']).reset_index().drop(columns=['index'])
    df_sampleid=pd.DataFrame(sampleid_list, columns=['sampleid']).reset_index().drop(columns=['index'])
    df_scanstring=pd.DataFrame(scanstring_list, columns=['XAS']).reset_index().drop(columns=['index'])
    df_angle=pd.DataFrame(angle_list, columns=['angle']).reset_index().drop(columns=['index'])
    df_xyzr= pd.concat([df_samplename,df_sampleid,df_angle,df_scanstring,df_sn,df_x,df_y,df_z,df_r,df_mono132,df_gap,df_epu,df_m0pit,df_m1pit,df_sgmensw,df_sgmensz,df_sgmhorz,df_exsv_2,df_epoch,df_epoch0,df_scancmd,df_colnames], axis=1)

    #df_xyzr['S']=df_xyzr['S'].astype(int)
    df_xyzr['E']=_experiment
    df_xyzr['F']=df_xyzr.index+1
    #df_xyzr['angle']=round(df_xyzr['R']%90,0)
    #print(df_xyzr)
    outfile=home+_experiment+'/pandas/df_xyzr.csv'
    df_xyzr.to_csv(outfile)
#SGMENSW, EXSV_2, SGMHORZ, GAP, EPU,
```
